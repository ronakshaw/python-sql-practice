# Join, merge & concatenate

```python
import pandas as pd
```

```python
df1 = pd.read_csv(r'/Users/ronak/Downloads/LOTR.csv')
df1
```

```python
df2 = pd.read_csv(r'/Users/ronak/Downloads/LOTR 2.csv')
df2
```

```python
df1.merge(df2)
```

```python
df1.merge(df2,how='inner',on=['FellowshipID'])
```

```python
df1.merge(df2,how='inner',on=['FellowshipID','FirstName'])
```

```python
df1.merge(df2,how='outer')
```

```python
df1.merge(df2,how='left')
```

```python
df1.merge(df2,how='right')
```

```python
df1.join(df2, on='FellowshipID', how='outer', lsuffix='_left', rsuffix='_right') #join will work good if indexing is done
```

```python
df4 = df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'), lsuffix='_left', rsuffix='_right',how='outer')
df4
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], join = 'outer', axis = 0)
```

```python
pd.concat([df1,df2], join = 'outer', axis = 1)
```
