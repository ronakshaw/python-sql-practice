# Filtering & ordering

```python
import pandas as pd
```

```python
df = pd.read_csv(r'/Users/ronak/Downloads/world_population.csv')
df
```

```python
df[df['Rank'] <= 10] # [] = give me, () = do this
```

```python
df[df['Country'] == 'India']
```

```python
df[df['Country'].isin(['India','China'])]
```

```python
specific_countries = ['India','China']
df[df['Country'].isin(specific_countries)]
```

```python
df[(df['Continent'] == 'Asia') & (df['Rank'] <= 10) ]
```

```python
df[(df['Continent'] == 'Asia') | (df['Continent'] == 'Africa')]
```

```python
df[df['Country'].str.contains('United')]
```

```python
df[df['Country'].str.startswith('Af')]
```

```python
df[df['Country'].str.len() >= 25]
```

```python
df[df['Country'].str.lower() == 'china'] # case-insensitive trick, for upper use .upper
```

```python
df2 = df.set_index('Country')
df2
```

```python
df2.filter(items = ['Continent','CCA3'], axis=1) # axis 1 for column and 0 for row
```

```python
df2[['Continent','CCA3']] # filter and [[]] both are same but we can use axis in filter
```

```python
df2.filter(like='united',axis=1)
```

```python
df2.loc['United States']
```

```python
df2.iloc[4]
```

```python
df[df['Rank'] < 10].sort_values(by='Rank',ascending = False) #ascending = false means descending
```

```python
df2[df2['Continent'] == 'Asia'] \
  .sort_values('2022 Population', ascending=False) \
  .head(5)
```
